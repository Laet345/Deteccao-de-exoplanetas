```python
#impoprtando os dados
from pathlib import Path

import pandas as pd

caminho_dados = Path("data/processed/wdbc_limpo.csv")
if not caminho_dados.exists():
    caminho_dados = Path("../data/processed/wdbc_limpo.csv")
if not caminho_dados.exists():
    raise FileNotFoundError(
        "Não encontrei data/processed/wdbc_limpo.csv. "
        "Execute o notebook a partir da pasta do projeto ou de notebooks."
    )

df = pd.read_csv(caminho_dados)

#separando em X e y
X = df.drop(columns=["id", "diagnosis"])
y = df["diagnosis"]

print(f"Dimensões de X: {X.shape}")
print(f"Dimensões de y: {y.shape}")
print("Diagnóstico (0 = benigno, 1 = maligno):")
display(y.value_counts().sort_index())
display(X.head())
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

# Use um pipeline com padronização para modelos sensíveis à escala,
# como regressão logística, KNN e SVM. O scaler aprende os parâmetros
# somente ao ajustar o pipeline com X_train, evitando vazamento para o teste.
pipeline_com_escala = Pipeline([
    ("scaler", StandardScaler()),
    ("modelo", LogisticRegression(max_iter=1000)),
])

# Para modelos baseados em árvores, como Random Forest e Decision Tree,
# a padronização geralmente não é necessária.
pipeline_sem_escala = Pipeline([
    ("modelo", RandomForestClassifier(random_state=42)),
])
```

```python
#dividindo estratificadamente (ne sei se existe essa palavra) em treino e teste
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)
```
